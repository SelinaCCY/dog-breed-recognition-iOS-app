```python
from bs4 import BeautifulSoup
import json
import requests
```

```python
# parse html content 
url = "https://www.thekennelclub.org.uk/search/breeds-a-to-z/"

response = requests.get(url) 
all_breed_data = [] 
```

```python
# code adapted from https://www.geeksforgeeks.org/implementing-web-scraping-python-beautiful-soup/
if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html.parser')
    # find all breed cards
    breed_cards = soup.find_all('div', class_='m-breed-card')

    for card in breed_cards:
        # extract breed name
        breed_name = card.find('strong', class_='m-breed-card__title').text.strip()
        
        # extract breed URL
        breed_url = card.find('a', class_='m-breed-card__link')['href']
        full_breed_url = f"https://www.thekennelclub.org.uk{breed_url}"
        
        # extract image URL
        image_url = card.find('img', class_='a-media__image')['src']

        # extract breed characteristics
        characteristics = {}
        for item in card.select('.m-breed-summary__item'):
            key = item.find('span', class_='m-breed-summary__key-label').text.strip()
            value = item.find('dd', class_='m-breed-summary__value').text.strip()
            characteristics[key] = value

        # combine all extracted data
        breed_data = {
            'Breed Name': breed_name,
            'Breed URL': full_breed_url,
            'Image URL': image_url,
            'Characteristics': characteristics
        }

        all_breed_data.append(breed_data)
else:
    print(f"Failed to retrieve content, status code: {response.status_code}")
```

```python
# convert the list of all breed data to JSON
# code adapted from https://www.freecodecamp.org/news/how-to-use-the-json-module-in-python/
all_breeds_json = json.dumps(all_breed_data, indent=4)
print(all_breeds_json)  # Print the JSON data
```

```python
# save the JSON data to a file
with open('breed_data.json', 'w') as file:
    file.write(all_breeds_json)
```
